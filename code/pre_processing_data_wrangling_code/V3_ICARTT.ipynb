```python
# Function to process the input file
def convert_to_csv(input_file, output_file):
    with open(input_file, 'r') as infile:
        lines = infile.readlines()

    # Extract headers (starting after metadata and until the "=====" line)
    header_start = None
    header_end = None
    for i, line in enumerate(lines):
        if line.strip() == "JDAY,  days":
            header_start = i
        if line.strip().startswith("==="):
            header_end = i
            break

    # Parse the header lines and format them
    headers = []
    if header_start is not None and header_end is not None:
        for line in lines[header_start:header_end]:
            parts = line.split(",")
            headers.append(parts[0].strip() + f" ({parts[1].strip()})" if len(parts) > 1 else parts[0].strip())

    # Extract data rows (starting after the metadata sections)
    data_start = header_end + 1
    data_rows = [line.strip() for line in lines[data_start:] if line.strip() and not line.startswith("=")]

    # Write to CSV
    with open(output_file, 'w', newline='') as csvfile:
        csv_writer = csv.writer(csvfile)
        csv_writer.writerow(headers)
        for row in data_rows:
            csv_writer.writerow(row.split(","))
    print(f"CSV file generated: {output_file}")

# Input and output file paths
input_file = r"D:\PVOCAL\Icartts\GTE\TRACEP\final-v4-mrgucgc_d_all.txt"  # Replace with your input file name
output_file = r"D:\PVOCAL\Other_Campaigns\old_coots\TRACEP_DC8.csv"  # Replace with your desired output file name

# Convert the file
convert_to_csv(input_file, output_file)
```

```python
def icartt_to_csv(icartt_filepath, csv_filepath):
    with open(icartt_filepath, 'r') as icartt_file:
        lines = icartt_file.readlines()

    # First line contains number of header lines and file format index
    header_line_count = int(lines[0].split(',')[0])

    # Extract header
    header = lines[:header_line_count]
    
    # Extract the variable names (last line of the header, usually right before the data)
    variable_names_line = header[-1].strip()
    variable_names = [name.strip() for name in variable_names_line.split(',')]

    # Extract data section (after the header)
    data_lines = lines[header_line_count:]

    # Create parent directory if it doesn't exist
    parent_dir = os.path.dirname(csv_filepath)
    os.makedirs(parent_dir, exist_ok=True)

    # Prepare to write CSV
    with open(csv_filepath, 'w', newline='') as csv_file:
        csv_writer = csv.writer(csv_file)
        
        # Write the column headers (variable names)
        csv_writer.writerow(variable_names)
        
        # Write the data rows
        for line in data_lines:
            data = line.strip().split(',')
            csv_writer.writerow(data)

    print(f"ICARTT file converted to CSV and saved as {csv_filepath}")
```

```python
def pemwest(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()

    # Extract the header
    header_line = lines[0].strip()
    headers = header_line.split()

    # Extract data rows
    data_lines = [line.strip().split() for line in lines[1:]]
    
    # Validate row lengths
    valid_data_lines = [row for row in data_lines if len(row) == len(headers)]

    # Log any problematic rows
    invalid_data_lines = [row for row in data_lines if len(row) != len(headers)]
    if invalid_data_lines:
        print(f"Warning: {len(invalid_data_lines)} rows had mismatched columns and were skipped.")

    # Create DataFrame
    df = pd.DataFrame(valid_data_lines, columns=headers)

    # Convert numeric columns
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='ignore')

    return df

# File path to the text file
file_path = "D:\PVOCAL\Icartts\GTE\PEMWest\A\pwa_all_hyd_m.txt"

# Parse the file
dataframe = pemwest(file_path)

# Display the DataFrame
print(dataframe)

# Save to a CSV file for easier viewing
dataframe.to_csv("parsed_data.csv", index=False)
```

```python
# This code is the seperator if the file has spaces instead of commas as the delimiter
def extract_date_from_filename(filename):
    """Extracts the date from the filename in the format 'YYYYMMDD'."""
    match = re.search(r'\d{8}', filename)
    if match:
        return pd.to_datetime(match.group(), format='%Y%m%d')
    else:
        raise ValueError(f"No date found in filename: {filename}")


def intexb_model_icartt_to_csv(icartt_filepath, csv_filepath):
    try:
        # Attempt to read the file using utf-8
        with open(icartt_filepath, 'r', encoding='utf-8') as icartt_file:
            lines = icartt_file.readlines()
    except UnicodeDecodeError:
        # Fallback to ISO-8859-1 encoding if utf-8 fails
        print(f"UTF-8 failed for {icartt_filepath}, trying ISO-8859-1 encoding...")
        with open(icartt_filepath, 'r', encoding='ISO-8859-1') as icartt_file:
            lines = icartt_file.readlines()

    # Extract number of header lines from the first line
    header_line_count = int(lines[0].split()[0])
    
    # Extract the header lines
    header = lines[:header_line_count]
    
    # Extract variable names (last header line)
    variable_names_line = header[-1].strip()
    variable_names = [name.strip() for name in variable_names_line.split()]
    
    # Extract data lines (everything after the header)
    data_lines = lines[header_line_count:]
    
    # Check if 'UTC' column exists, which we need for calculating Datetime
    if "UTC" not in variable_names:
        raise ValueError(f"'UTC' variable not found in the ICARTT file: {icartt_filepath}")
    
    # Extract the base date from the filename (YYYYMMDD format)
    base_date = extract_date_from_filename(icartt_filepath)

    # Add the "Datetime" column to the variable names
    variable_names.append("Datetime")

    # Prepare to write to CSV
    with open(csv_filepath, 'w', newline='') as csv_file:
        csv_writer = csv.writer(csv_file)
        
        # Write the column headers
        csv_writer.writerow(variable_names)
        
        # Write the data rows
        for line in data_lines:
            # Split data using whitespace
            data = line.strip().split()

            # Get UTC value (elapsed seconds)
            utc_seconds = float(data[variable_names.index("UTC")])
            
            # Calculate the full Datetime
            full_datetime = base_date + timedelta(seconds=utc_seconds)
            
            # Append the calculated datetime to the data
            data.append(full_datetime.strftime("%Y-%m-%d %H:%M:%S"))
            
            # Write the row with the new Datetime column
            csv_writer.writerow(data)
    
    print(f"ICARTT file converted to CSV with Datetime column and saved as {csv_filepath}")

def convert_all_indexb_model_icartt_to_csv(directory_path):
    # List all .ict files in the directory
    ict_files = [f for f in os.listdir(directory_path) if f.endswith('.ict')]
    
    if not ict_files:
        print("No .ict files found in the directory.")
        return
    
    for ict_file in ict_files:
        icartt_filepath = os.path.join(directory_path, ict_file)
        # Define output CSV file path (same directory, change extension to .csv)
        csv_filename = os.path.splitext(ict_file)[0] + '.csv'
        csv_filepath = os.path.join(directory_path, csv_filename)

        try:
            # Convert each file
            intexb_model_icartt_to_csv(icartt_filepath, csv_filepath)
        except Exception as e:
            print(f"Error converting {ict_file}: {e}")

# Convert all files in the specified directory
convert_all_indexb_model_icartt_to_csv(r'D:\PVOCAL\model_data\intexna\RAQMS')
```

```python
def combine_csv_files(input_directory, output_file):
    # List all CSV files in the directory
    csv_files = [f for f in os.listdir(input_directory) if f.endswith('.csv')]
    
    if not csv_files:
        print("No CSV files found in the directory.")
        return
    
    # List to store all dataframes
    df_list = []
    
    # Read each CSV and append to the list
    for csv_file in csv_files:
        csv_path = os.path.join(input_directory, csv_file)
        try:
            df = pd.read_csv(csv_path)
            df_list.append(df)
            print(f"Added {csv_file} to the dataframe list.")
        except Exception as e:
            print(f"Error reading {csv_file}: {e}")
    
    # Concatenate all dataframes
    combined_df = pd.concat(df_list, ignore_index=True)
    
    # Write the combined dataframe to a new CSV file
    combined_df.to_csv(output_file, index=False)
    print(f"Combined CSV saved as {output_file}")

# Call the function with the input directory and output file path
combine_csv_files(r'D:\PVOCAL\model_data\intexna\RAQMS', r'D:\PVOCAL\model_data\intexna_RAQMS.csv')
```

```python
def extract_date_from_filename(filename):
    """Extracts the date from the filename in the format 'YYYYMMDD'."""
    match = re.search(r'\d{8}', filename)
    if match:
        return pd.to_datetime(match.group(), format='%Y%m%d')
    else:
        raise ValueError(f"No date found in filename: {filename}")

def combine_csv_files_with_datetime(input_directory, output_file):
    # List all CSV files in the directory
    csv_files = [f for f in os.listdir(input_directory) if f.endswith('.csv')]
    
    if not csv_files:
        print("No CSV files found in the directory.")
        return
    
    # List to store all dataframes
    df_list = []
    
    # Process each CSV file
    for csv_file in csv_files:
        csv_path = os.path.join(input_directory, csv_file)
        try:
            # Read the CSV file
            df = pd.read_csv(csv_path)
            
            # Extract the start date from the filename
            start_date = extract_date_from_filename(csv_file)
            
            # Convert 'Time_Start' to datetime if it exists
            if 'Time_Start' in df.columns:
                df['Datetime_Start'] = start_date + pd.to_timedelta(df['Time_Start'].fillna(0), unit='s')
                df['Datetime_Start'] = df['Datetime_Start'].where(df['Time_Start'].notna(), pd.NaT)
            
            # Convert 'Fractional_Day' to datetime if it exists
            if 'Fractional_Day' in df.columns:
                integer_day = df['Fractional_Day'].dropna().astype(int)
                fractional_day = df['Fractional_Day'].dropna() - integer_day
                date_part = pd.to_datetime(integer_day - 1, origin=start_date, unit='D')
                time_part = pd.to_timedelta(fractional_day, unit='D')
                df['Datetime_Fractional_Day'] = pd.NaT  # initialize with NaT
                df.loc[df['Fractional_Day'].notna(), 'Datetime_Fractional_Day'] = date_part + time_part
            
            # Add to the list of DataFrames
            df_list.append(df)
            print(f"Added {csv_file} to the dataframe list.")
        
        except Exception as e:
            print(f"Error processing {csv_file}: {e}")
    
    # Concatenate all dataframes
    combined_df = pd.concat(df_list, ignore_index=True)
    
    # Write the combined dataframe to a new CSV file
    combined_df.to_csv(output_file, index=False)
    print(f"Combined CSV saved as {output_file}")

# Call the function with the input directory and output file path
combine_csv_files_with_datetime(r'D:\PVOCAL\Other_Campaigns\old_coots\temp', r'D:\PVOCAL\Other_Campaigns\INTEXA.csv')
```
